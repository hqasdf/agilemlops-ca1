```python
import mlflow

mlflow.set_tracking_uri(
    "http://mlflow-aws-eb.eba-9bnrxizu.us-east-1.elasticbeanstalk.com"
)
mlflow.set_registry_uri(
    "http://mlflow-aws-eb.eba-9bnrxizu.us-east-1.elasticbeanstalk.com"
)
mlflow.set_experiment("PBI05-connection-check")

class TinyTestModel(mlflow.pyfunc.PythonModel):
    def predict(self, context, model_input):
        return model_input

with mlflow.start_run(run_name="Suresh Nithish-registry-check"):
    model_info = mlflow.pyfunc.log_model(
        name="pbi05_dummy_model",
        python_model=TinyTestModel(),
        registered_model_name="pbi05_registry_smoke_test",
    )

print("Registered test model:", model_info.model_uri)
```
